```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()
print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Levels:", digits.target_names)

plt.figure(figsize=(4,4))
plt.imshow(digits.images[0], cmap='pink')
plt.title(f"Label:{digits.target[99]}")
plt.axis('off')
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt          # "maplotib" -> "matplotlib"

digits = load_digits()
fig, axes = plt.subplots(2, 5, figsize=(10, 5))

for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='gray')   # "Images" -> "images", "cmp" -> "cmap"
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')

plt.show()   # call once, outside the loop, on plt not ax
```
